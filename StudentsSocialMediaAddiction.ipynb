```python
import pandas as pd
from google.colab import files
import os


def load_data():
    """
    Loads the dataset from the CSV file uploaded to the Colab environment.
    """

    file_name = "/content/Students Social Media Addiction.csv"

    if not os.path.exists(file_name):
        print(f"❌ ERROR: File '{file_name}' not found.")
        print("Please make sure that you have uploaded the file.")
        return pd.DataFrame()

    try:

        df = pd.read_csv(file_name)


        required_cols = ['Avg_Daily_Usage_Hours', 'Addicted_Score', 'Conflicts_Over_Social_Media', 'Mental_Health_Score']

        if all(col in df.columns for col in required_cols):
            print(f"✅ Dataset '{file_name}' successfully loaded. Total respondents: {len(df)}")
            print("-" * 50)
            return df
        else:
            print("❌ ERROR: The loaded file is missing one or more required columns.")
            return pd.DataFrame()

    except Exception as e:
        print(f"❌ An unexpected error occurred during loading: {e}")
        return pd.DataFrame()

def inference_engine(respondent_data):
    """
    The core function acting as the Inference Engine.
    Applies 'If-Then' rules to the input data (facts) to derive a conclusion (inference).
    """

    usage = respondent_data['Avg_Daily_Usage_Hours']
    anxiety = respondent_data['Addicted_Score']
    conflict = respondent_data['Conflicts_Over_Social_Media']
    satisfaction = respondent_data['Mental_Health_Score']


    risk_level = "MODERATE"
    recommendation = "General awareness and monitoring advised."


    if (usage >= 4.0) and (anxiety >= 4) and (conflict >= 4):
        risk_level = "**CRITICAL ADDICTION RISK**"
        recommendation = "IMMEDIATE PROFESSIONAL SUPPORT REQUIRED. Consider a complete digital detox period and seek specialized counselling."


    elif (satisfaction <= 2) and (conflict >= 4):
        risk_level = "HIGH RELATIONSHIP STRESS RISK"
        recommendation = "Focus on non-SM interactions. Establish 'Social Media Free Zones' with partners to rebuild connection."


    elif (usage >= 3.0) and (anxiety >= 3):
        risk_level = "ELEVATED USAGE RISK"
        recommendation = "Implement active time management strategies. Utilize SM app limitations and disable non-essential notifications."


    elif (usage < 2.0) and (anxiety < 2) and (satisfaction >= 4):
        risk_level = "LOW RISK"
        recommendation = "Maintain current healthy usage habits. Leverage social media for professional/educational growth."

    return risk_level, recommendation

df_analysis = load_data()

if not df_analysis.empty:

    df_analysis[['Risk_Level', 'Advisory_Recommendation']] = df_analysis.apply(
        lambda row: inference_engine(row),
        axis=1,
        result_type='expand'
    )


    print("\n📊 EXPERT SYSTEM DIAGNOSTIC REPORT (Inference Results):")
    print("-" * 50)


    print(df_analysis)

    print("-" * 50)
    print(f"\nReport Generated for {len(df_analysis)} respondents.")

```
